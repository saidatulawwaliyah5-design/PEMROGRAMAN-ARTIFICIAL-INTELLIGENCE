```python
# Program Menentukan Jenis Segitiga

x = int(input("Masukkan sisi x: "))
y = int(input("Masukkan sisi y: "))
z = int(input("Masukkan sisi z: "))

if x == y == z:
    print("Segitiga Sama Sisi (Equilateral Triangle)")

elif x == y or x == z or y == z:
    print("Segitiga Sama Kaki (Isosceles Triangle)")

else:
    print("Segitiga Sembarang (Scalene Triangle)")
```

```python
# Program Menentukan Zodiak Cina (Shio)

tahun = int(input("Masukkan tahun lahir: "))

if tahun % 12 == 4:
    print("Shio Anda adalah Tikus")

elif tahun % 12 == 5:
    print("Shio Anda adalah Kerbau")

elif tahun % 12 == 6:
    print("Shio Anda adalah Macan")

elif tahun % 12 == 7:
    print("Shio Anda adalah Kelinci")

elif tahun % 12 == 8:
    print("Shio Anda adalah Naga")

elif tahun % 12 == 9:
    print("Shio Anda adalah Ular")

elif tahun % 12 == 10:
    print("Shio Anda adalah Kuda")

elif tahun % 12 == 11:
    print("Shio Anda adalah Kambing")

elif tahun % 12 == 0:
    print("Shio Anda adalah Monyet")

elif tahun % 12 == 1:
    print("Shio Anda adalah Ayam")

elif tahun % 12 == 2:
    print("Shio Anda adalah Anjing")

else:
    print("Shio Anda adalah Babi")
```
