```python
i = 2
while(i < 100):
    j = 2
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, " is prime")
    i = i + 1

print("Good bye!")
```

```python
angka = [12, 7, 45, 28, 33, 50, 9, 64, 81, 100, 15, 22]

genap = []
ganjil = []

for n in angka:
    if n % 2 == 0:
        genap.append(n)
    else:
        ganjil.append(n)

print("List awal      :", angka)
print("Bilangan genap :", genap)
print("Bilangan ganjil:", ganjil)
print()
print("Banyak bilangan genap  :", len(genap))
print("Banyak bilangan ganjil :", len(ganjil))
print("Jumlah bilangan genap  :", sum(genap))
print("Jumlah bilangan ganjil :", sum(ganjil))
```

```python
n = int(input("Masukkan banyak deret Fibonacci: "))

a, b = 0, 1
hitung = 0
print("Deret Fibonacci:")
while hitung < n:
    print(a, end=" ")
    a, b = b, a + b
    hitung += 1
print()
```
