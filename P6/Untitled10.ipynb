```python
x, y = 20, 10
print("nilai awal x= %d, y= %d" % (x, y))
temp = x      # variabel sementara
x = y         # tukar nilai
y = temp
print("nilai x setelah di tukar: %d" % (x))
print("nilai y setelah di tukar: %d" % (y))
```

```python
a = 0b10100   # 20
b = 0b01001   # 9
bitwise_and = a & b
bitwise_or = a | b
bitwise_xor = a ^ b
print("Nilai dari operasi bitwise \"AND\" antara a:%d, b:%d -> (a and b) = %d" % (a, b, bitwise_and))
print("Nilai dari operasi bitwise \"OR\" antara a:%d, b:%d -> (a or b) = %d" % (a, b, bitwise_or))
print("Nilai dari operasi bitwise \"XOR\" antara a:%d, b:%d -> (a xor b) = %d" % (a, b, bitwise_xor))
```

```python
num = 1234
x = num // 1000
x1 = (num - x*1000) // 100
x2 = (num - x*1000 - x1*100) // 10
x3 = num - x*1000 - x1*100 - x2*10
print("angka = ", num)
print("Jumlah dari angka = ", x + x1 + x2 + x3)
```
